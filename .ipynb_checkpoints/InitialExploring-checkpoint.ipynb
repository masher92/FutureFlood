### Add a column categorising the time period

```python
def categorize_year(year):
    if year < 2020:
        return "Baseline"
    elif year < 2050:
        return "Present"
    else:
        return "Future"

rainfall['time_period'] = rainfall['start_year'].apply(categorize_year)
```

## Seasonal plot

```python
rainfall['date'] = pd.to_datetime(dict(year=rainfall['start_year'], month=rainfall['start_month'],day=rainfall['start_day']))
rainfall['day_of_year'] = rainfall['date'].dt.dayofyear
rainfall['theta'] = 2 * np.pi * (rainfall['day_of_year'] - 1) / 365
```

```python
np.unique(rainfall['start_month'], return_counts=True)
```

```python
def calc_mean_day_and_dispersion(theta):
    
    x = np.cos(theta)
    y = np.sin(theta)

    x_mean = np.mean(x)
    y_mean = np.mean(y)
    
    if x_mean <= 0:
        D_mean = (np.arctan(y_mean/x_mean) + np.pi) * 365.25/(2*np.pi)
    elif x_mean > 0 and y_mean >= 0:
        D_mean = np.arctan(y_mean/x_mean) * 365.25/(2*np.pi)
    elif x_mean > 0 and y_mean < 0:
        D_mean = (np.arctan(y_mean/x_mean) + 2*np.pi) * 365.25/(2*np.pi)
    
    R = np.sqrt(x_mean**2 + y_mean**2)

    return D_mean, R

D_mean, R = calc_mean_day_and_dispersion(rainfall['theta'])
R
```

### Plot whether soil moisture changes in the future

```python
# Define colors for each category
colors = {"Baseline": "blue", "Present": "green", "Future": "red"}

plt.figure(figsize=(8, 6))

for period in ["Baseline", "Present", "Future"]:
    subset = rainfall[rainfall['time_period'] == period]
    plt.hist(subset['mean_soil_moisture'], bins=9, alpha=0.8, 
             label=period, color=colors[period])

plt.xlabel("Mean Soil Moisture")
plt.ylabel("Frequency")
plt.title("Distribution of Mean Soil Moisture by Time Period")
plt.legend()
plt.show()
```

```python
mean_soil_moistures = []
for file in os.listdir(home_dir + "kv25483/FutureFlood/Data/NewcastleExample/soil/Ens01_23/"):
    with rasterio.open(home_dir + f"kv25483/FutureFlood/Data/NewcastleExample/soil/Ens01_23/{file}") as src:
        # print(src.width, src.height)
        # print(src.count)  # number of bands
        img = src.read(1)  # read first band
        # Replace -9999 (or src.nodata) with np.nan
        img = img.astype(float)  # ensure it's float, so np.nan works
        img[img == -9999] = np.nan
        
        # Now compute the mean, ignoring NaNs
        mean_soil_moisture = round(np.nanmean(img),2)
        print("Average soil moisture:", mean_soil_moisture)

        # Append to list
        mean_soil_moistures.append(mean_soil_moisture)
```

```python
newcastle_proj = newcastle.to_crs(epsg=27700)

from shapely.geometry import Point

# Extract coordinates
x = soil_moisture_test.coord('projection_x_coordinate').points
y = soil_moisture_test.coord('projection_y_coordinate').points

# Create 2D meshgrid of coordinates
xx, yy = np.meshgrid(x, y)

# Flatten coordinates
points = np.vstack((xx.ravel(), yy.ravel())).T

# Get catchment geometry
catchment_geom = newcastle_proj.geometry.iloc[0]

# Test which grid cells are inside catchment
mask = np.array([catchment_geom.contains(Point(pt)) for pt in points])

# Reshape to grid
mask = mask.reshape(len(y), len(x))
```

```python
data = first_hour.data.astype(float)

# Mask outside catchment
data[~mask] = np.nan

first_hour.data = data

plt.figure(figsize=(8, 6))
mesh = iplt.pcolormesh(first_hour)

# Get catchment bounds
xmin, ymin, xmax, ymax = newcastle.total_bounds

ax = plt.gca()
ax.set_xlim(xmin, xmax)
ax.set_ylim(ymin, ymax)

newcastle.boundary.plot(ax=ax, edgecolor="red", linewidth=2)

plt.colorbar(mesh, label="Soil Moisture")
plt.title("Soil Moisture – First Hour (Newcastle Catchment)")
plt.tight_layout()
plt.show()
```

## Test relationships between variables

```python
x = rainfall['mean_soil_moisture']
y = rainfall['flood_area_km2_10cm']

slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
  return slope * x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x, y, color='black')
plt.plot(x, mymodel)
plt.show()
```
