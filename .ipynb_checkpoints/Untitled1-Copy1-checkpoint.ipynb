```python
def set_osgb_projection(cube):
    """Assign OSGB coordinate system to x/y coordinates."""
    bng = iris.coord_systems.OSGB()
    cube.coord('projection_x_coordinate').coord_system = bng
    cube.coord('projection_y_coordinate').coord_system = bng
    return cube

def guess_bounds_if_missing(cube):
    """Guess bounds on x/y coordinates if not already present."""
    for coord_name in ['projection_x_coordinate', 'projection_y_coordinate']:
        coord = cube.coord(coord_name)
        if not coord.has_bounds():
            coord.guess_bounds()
    return cube

def prepare_flood_cube(cube, catchment_poly, plot=False):
    """Apply standard transformations to a flood cube."""
    cube = set_osgb_projection(cube)
    cube = guess_bounds_if_missing(cube)
    cube = filter_closer_to_catchment(cube, catchment_poly, plot=plot)
    return cube
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Extract values
x = results_df['max_precip'].values
y = results_df['flooded_area'].values

# Avoid log(0) by replacing zeros with a small number
y_safe = np.where(y == 0, 1e-3, y)

plt.figure(figsize=(6,6))
plt.scatter(x, y_safe, alpha=0.7)

# 1:1 reference line (in original units)
min_val = min(x.min(), y_safe.min())
max_val = max(x.max(), y_safe.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', label='1:1 line')

plt.xlabel("Max precipitation (mm/hr)")
plt.ylabel("Flooded area (km²)")
plt.yscale('log')   # Log scale for flooded area
plt.ylim(1e-3, max_val*1.1)  # Adjust limits to include very small values
plt.legend()
plt.title("Max precipitation vs Flooded area (log scale)")
plt.grid(True, which="both", ls="--", lw=0.5)
plt.show()
```
