### Check plotting with boundaries

```python
hc_5km = xr.open_dataset("../../Data/HydraulicConductivity/GB_5km.nc")
```

```python
n_classes = len(quantiles) + 1

# Auto-generate labels from bin edges (reversed to match flipped class order)
edges = [q0] + quantiles + [np.inf]
class_labels = [f"{edges[i]}–{edges[i+1]}" for i in range(len(edges)-1)][::-1]

# Auto-generate colors from a colormap, sampled evenly across n_classes
base_cmap = plt.get_cmap('viridis')  # swap for 'RdYlBu_r', 'plasma', etc. as preferred
colors = [base_cmap(i) for i in np.linspace(0, 1, n_classes)]

# Get data
hk_vals = hc_5km['aggregated_mean']
if np.ma.is_masked(hk_vals):
    hk_vals = hk_vals.filled(np.nan)

# Convert data to classes (flip the order), bins built from the quantiles list
bins = [q0] + quantiles
classified = (n_classes + 1) - np.digitize(hk_vals, bins=bins)
classified = classified.astype(float)
classified[np.isnan(hk_vals)] = np.nan

# Assign back to cube
hc_5km["classified"] = (("y", "x"), classified)

fig, ax = plt.subplots(figsize=(4, 5))
cmap = ListedColormap(colors)
bounds = np.arange(0.5, n_classes + 1.5, 1)
norm = BoundaryNorm(bounds, cmap.N)
mesh = hc_5km["classified"].plot(ax=ax,cmap=cmap,norm=norm,add_colorbar=False)
ticks = np.arange(1, n_classes + 1)
cbar = fig.colorbar(mesh,ax=ax,orientation="vertical",fraction=0.046,pad=0.04,ticks=ticks)
cbar.ax.set_yticklabels(class_labels)
cbar.set_label("HC Class")
plt.show()
```

### Check relationships between rainfall and flooding in each class

```python
intensity_vars = ['max_precip', 'total_acc']
shape_vars = ['peak_position_ratio', 'd50', 'gini', 'time_based_std', 'dry_ratio', 'max_quartile_index']
neighbourhood_vars = ['cluster_rain_total_t60', 'neighbourhood_rain_total', 'neighbourhood_rain_sum_excl_peak',
                     'cluster_rain_total_t50', "cluster_rain_total_t80"] # ['cluster_flood_10_area_t80']
antecedent_vars = ['mean_sm_2_before_event']
length_vars = ['length']
candidate_vars = length_vars + antecedent_vars + neighbourhood_vars + shape_vars + intensity_vars
```

```python
print(len(rainfall_events_all_df))
rainfall_events_all_df = rainfall_events_all_df[rainfall_events_all_df['10cm_area']>0].copy()
print(len(rainfall_events_all_df))
```

```python
import pandas as pd
import statsmodels.formula.api as smf
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score

df = rainfall_events_all_df

results = []

for cls in sorted(df["HC_status_2"].dropna().unique()):

    sub = df[df["HC_status_2"] == cls].dropna(subset=candidate_vars + ["10cm_area"])

    print(f"\n--- HC class {cls} (n={len(sub)}) ---")

    # ---------------------------------------------------------
    # OLS models
    # ---------------------------------------------------------

    ols_models = [("Baseline (intensity+acc)", intensity_vars),
        ("+ antecedent", intensity_vars + antecedent_vars),
        ("+ neighbourhood", intensity_vars + antecedent_vars + neighbourhood_vars),
        ("+ shape/temporal-loading",
         intensity_vars + antecedent_vars + neighbourhood_vars + shape_vars),]

    for label, predictors in ols_models:

        formula = 'Q("10cm_area") ~ ' + " + ".join(predictors)

        model = smf.ols(formula, data=sub).fit()

        r2 = model.rsquared
        adj_r2 = model.rsquared_adj

        print(f"  {label}: R² = {r2:.3f}, "f"Adj. R² = {adj_r2:.3f}")

        results.append({ "HC_class": cls, "n": len(sub), "model": label, "R2": r2, "Adjusted_R2": adj_r2, 
                        "CV_R2_mean": None, "CV_R2_std": None})

    # ---------------------------------------------------------
    # Random Forest
    # ---------------------------------------------------------        
        
    rf_models = [("Baseline (intensity+acc)", intensity_vars),
        ("+ antecedent", intensity_vars + antecedent_vars),
        ("+ neighbourhood",
         intensity_vars + antecedent_vars + neighbourhood_vars),
        ("+ shape/temporal-loading",
         intensity_vars + antecedent_vars + neighbourhood_vars + shape_vars),]

    for label, predictors in rf_models:

        X = sub[predictors]
        y = sub["10cm_area"]

        rf = RandomForestRegressor(n_estimators=300,max_depth=8,random_state=0,n_jobs=-1)

        scores = cross_val_score(rf,X,y,cv=5,scoring="r2")

        mean_r2 = scores.mean()
        std_r2 = scores.std()

        print(f"  RF {label}: "f"R² = {mean_r2:.3f} +/- {std_r2:.3f}")

        results.append({"HC_class": cls,"n": len(sub),"model": f"Random Forest: {label}","R2": None,
                        "Adjusted_R2": None,"CV_R2_mean": mean_r2,"CV_R2_std": std_r2})
        
# -------------------------------------------------------------
# Results dataframe
# -------------------------------------------------------------

results_df = pd.DataFrame(results)

print("\n\n================ RESULTS ================\n")
print(results_df.to_string(index=False))


# -------------------------------------------------------------
# Save to CSV
# -------------------------------------------------------------

output_file = "HC_class_model_results.csv"

results_df.to_csv(output_file, index=False)

print(f"\nResults saved to: {output_file}")
```

```python
df= rainfall_events_all_df
def run_stage(sub, predictors, label):
    formula = 'Q("10cm_area") ~ ' + ' + '.join(predictors)
    model = smf.ols(formula, data=sub).fit()
    # print(model.summary())
    print(f"  {label}: R² = {model.rsquared:.3f}")
    return model

results = {}
for cls in sorted(df['HC_status_2'].unique()):
    sub = df[df['HC_status_2'] == cls].dropna(subset=candidate_vars + ['10cm_area'])
    print(f"\n--- HC class {cls} (n={len(sub)}) ---")
    
    m1 = run_stage(sub, intensity_vars , "Baseline (intensity+acc)")
    m2 = run_stage(sub, intensity_vars + antecedent_vars, "+antecedent")
    m3 = run_stage(sub, intensity_vars + antecedent_vars + neighbourhood_vars, "+ neighbourhood")
    m4 = run_stage(sub, intensity_vars + antecedent_vars + neighbourhood_vars + shape_vars, "+ shape/temporal-loading")
    
    results[cls] = {'baseline': m1, 'shape': m2, 'length': m3, 'full': m4}
    
    
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score

for cls in sorted(df['HC_status_2'].unique()):
    sub = df[df['HC_status_2'] == cls].dropna(subset=candidate_vars + ['10cm_area'])
    X = sub[candidate_vars]
    y = sub['10cm_area']
    rf = RandomForestRegressor(n_estimators=300, max_depth=8, random_state=0, n_jobs=-1)
    scores = cross_val_score(rf, X, y, cv=5, scoring='r2')
    print(f"{cls}: RF CV R² = {scores.mean():.3f} (+/- {scores.std():.3f})")
```

### Partial dependence testing

Left panel): partial dependence of 10cm_area on hc_at_peak shows a sharp drop between roughly HC = 0.15 and HC = 0.35, then flattens out into a long, gently-declining tail from ~0.4 to 1.75. This means most of the actual "information" the model is using is concentrated in that low-HC region, and beyond ~0.4 the marginal effect of HC is small.

Right panel (2D interaction with max_precip): the contours confirm this isn't a clean, precip-independent threshold — the transition zone (the tightly-packed contour lines) sits at low hc_at_peak across all precip levels, but the steepness of the gradient changes with precip. At high max_precip (>55), the "0.61 vs 0.51" bands are compressed into a narrow low-HC strip, suggesting the model finds HC most discriminating precisely when rainfall is intense. At lower precip the bands are more evenly spread across HC, meaning HC matters less when rainfall itself isn't extreme.

```python
from sklearn.inspection import PartialDependenceDisplay

rf = RandomForestRegressor(n_estimators=300, max_depth=8, random_state=0, n_jobs=-1)
X = df[candidate_vars + ['hc_at_peak']].dropna()
y = df.loc[X.index, '10cm_area']
rf.fit(X, y)

PartialDependenceDisplay.from_estimator(
rf, X,
    features=['hc_at_peak',
              ('hc_at_peak', 'max_precip')],  # 2D interaction plot
    kind='average')
```
