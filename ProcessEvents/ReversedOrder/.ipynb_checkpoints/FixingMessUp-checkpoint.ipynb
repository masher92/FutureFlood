```python
import shutil
import numpy as np
import pandas as pd

KEY = ['ens', 'hydro_year', 'event_num']
SERIES = [('rainpeak_rain_series', 'rainpeak_found'),
          ('floodpeak_rain_series', 'floodpeak_found')]

def fix_broadcast(df):
    """Collapse broadcast rows back to one row per event. Returns (df, n_events_fixed)."""
    # A broadcast row holds a scalar where a good row holds an array
    bad = df['rainpeak_found'].astype(bool) & df['rainpeak_rain_series'].map(lambda v: np.ndim(v) == 0)
    if not bad.any():
        return df, 0

    good_df = df[~bad]
    bad_df = df[bad]
    fixed = bad_df.drop_duplicates(subset=KEY, keep='first').copy()   # keeps original index

    for col, found_col in SERIES:
        if col not in bad_df.columns:
            continue
        # Rebuild each event's series from its rows, in row order
        d = {k: s.to_numpy(dtype=np.float32)
             for k, s in bad_df.groupby(KEY, sort=False)[col]}
        fixed[col] = pd.Series(
            [d[k] for k in zip(fixed['ens'], fixed['hydro_year'], fixed['event_num'])],
            index=fixed.index, dtype=object)
        # No flood peak -> there was no series, so restore NaN rather than a bogus array
        fixed.loc[~fixed[found_col].astype(bool), col] = np.nan

    out = pd.concat([good_df, fixed]).sort_index().reset_index(drop=True)
    assert not out.duplicated(subset=KEY).any(), "still duplicate events after fix"
    return out, len(fixed)


DRY_RUN = True   # set to False once the printout looks right

for TARGET_HA in list(CATCHMENT_LOOKUP_DICT.keys()):
    catchment_name = CATCHMENT_LOOKUP_DICT[TARGET_HA]

    old_fp = os.path.join(OUT_DIR, f"Catchment_{TARGET_HA}", f"{catchment_name}.pkl")
    try:
        old_df = pd.read_pickle(old_fp)
    except Exception:
        old_df = None   # reset so we never compare against the previous catchment
        print(f"{TARGET_HA} {catchment_name}: no old df")

    fp = os.path.join(OUT_DIR, f"Catchment_{TARGET_HA}", f"{catchment_name}_new_Nov.pkl")
    if not os.path.exists(fp):
        continue

    df = pd.read_pickle(fp)
    
    fixed_fp = os.path.join(OUT_DIR, f"Catchment_{TARGET_HA}", f"{catchment_name}_new_Nov_error.pkl")
    df.to_pickle(fp)
        
    fixed_df, n_fixed = fix_broadcast(df)

    match = old_df is not None and len(fixed_df) == len(old_df)
    status = "OK" if n_fixed == 0 else f"fixing {n_fixed} events, {len(df)} -> {len(fixed_df)} rows"
    print(f"{TARGET_HA} {catchment_name}: {status}, old rows: {len(old_df) if old_df is not None else 'n/a'}")

    if not match:
        print(f"   *** ERROR {TARGET_HA} {catchment_name}: fixed rows ({len(fixed_df)}) != old rows "
              f"({len(old_df) if old_df is not None else 'n/a'})")

    if n_fixed == 0:
        continue

    if not match:
        print(f"   *** NOT saving {catchment_name}")
        continue

    if not DRY_RUN:
        # shutil.copy(fp, fp.replace('.pkl', '_broadcast_backup.pkl'))
        fixed_df.to_pickle(fp)
```
