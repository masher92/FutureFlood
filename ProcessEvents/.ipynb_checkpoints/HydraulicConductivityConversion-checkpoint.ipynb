### Convert soil texture data to hydraulic conducitivity values

```python
# Get the raw numpy array
soil_values = soil_texture.values  # shape (2450, 2855)

# Create output array, defaulting to NaN
hk_values = np.full(soil_values.shape, np.nan)

# Map each USDA class to its hydraulic conductivity value
for usda_class, hk in lookup.items():
    hk_values[soil_values == usda_class] = hk

# Put back into an xarray DataArray, preserving spatial metadata
hk_raster = soil_texture.copy(data=hk_values)
hk_raster.name = "hydraulic_conductivity"
```

Step 1: Build hk_5km for all catchments

Step 2: Define hk bins once, upfront

```python
# Bin edges — adjust to match the range of your lookup values
# e.g. if lookup spans ~0.001 to 100 mm/hr, log bins make sense
bin_edges = [0, 0.01, 0.1, 1.0, 10.0, 100.0]  # adjust to your lookup range
bin_labels = ["very low", "low", "medium", "high", "very high"]

# Assign each 5km cell a bin index (0=very low ... 4=very high)
hk_binned = xr.apply_ufunc(
    np.digitize,
    hk_mosaic_5km,
    kwargs={"bins": bin_edges},
    dask="parallelized",
    output_dtypes=[int]
)
hk_binned.name = "hk_bin"
```

Step 3: Loop over years, compute mean SM per hk bin

```python
results = []

for year in tqdm(range(1980, 2081)):
    try:
        sm_file = glob.glob(
            f"{sm_dir}/r001i1p*****_{year-1}1201-{year}1130_mrso.nc"
        )[0]
    except IndexError:
        print(f"Missing: {year}")
        continue

    sm_xr = xr.DataArray.from_iris(iris.load(sm_file)[0])  # (time, y, x)

    # For each bin, mask SM and take spatial mean across all timesteps
    bin_means = {}
    masks = {label: hk_binned == (i + 1) for i, label in enumerate(bin_labels)}

    # Compute all bins in one go
    to_compute = {
        label: sm_xr.where(mask).mean(dim=["x", "y"])
        for label, mask in masks.items()
    }
    computed = xr.compute(*to_compute.values())  # single dask graph execution

    for label, result in zip(to_compute.keys(), computed):
        df = pd.DataFrame({
            "time": result.time.values,
            "mean_sm": result.values,
            "hk_bin": label,
            "year": year
        })
        results.append(df)

df_all = pd.concat(results, ignore_index=True)
df_all.to_parquet("sm_by_hk_bin_1980_2080.parquet")
```

Step 4: Plot SM over time by hk bin

```python
import matplotlib.pyplot as plt

df = pd.read_parquet("sm_by_hk_bin_1980_2080.parquet")

# Annual mean per bin
annual = df.groupby(["year", "hk_bin"])["mean_sm"].mean().reset_index()

fig, ax = plt.subplots(figsize=(14, 5))

colors = ["#2c7bb6", "#abd9e9", "#ffffbf", "#fdae61", "#d7191c"]
for color, label in zip(colors, bin_labels):
    subset = annual[annual["hk_bin"] == label]
    ax.plot(subset["year"], subset["mean_sm"], label=label, color=color)

ax.set_ylabel("Mean soil moisture (kg/m²)")
ax.set_xlabel("Year")
ax.set_title("Mean soil moisture by hydraulic conductivity class, 1980–2080")
ax.legend(title="HC class")
plt.tight_layout()
plt.show()
```
