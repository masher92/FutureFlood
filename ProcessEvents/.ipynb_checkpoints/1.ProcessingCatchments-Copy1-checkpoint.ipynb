```python
import numpy as np
import pandas as pd
import cftime
import re
import time
import gc
import os
import sys
import math
import glob
import datetime
from scipy.ndimage import label, generate_binary_structure

from functions import *
from functions_stage2 import (get_rainfall_cube_subsection, find_max_precip_location_new, maybe_diagnose, analyse_peak_event,
                             plot_cluster_check,get_rainfall_lookback_cube)
verbose=True

# ── Config ────────────────────────────────────────────────────────────────────
from config import RAINFALL_CSV_DIR, CATCHMENT_LOOKUP_DICT, OUT_DIR, CATCHMENTS, ENSEMBLE_MEMBERS 

def plot_sm_check(sm_data, sm_time_arr, sm_day_idx, anchor_key, yi, xi,
                  values, time_coord, n=10):
    idx = sm_day_idx[anchor_key]
    lo, hi = max(idx - n, 0), min(idx + n + 1, len(sm_time_arr))
    offsets = np.arange(lo, hi) - idx
    sm_series = sm_data[lo:hi, yi, xi]

    # daily rainfall at the peak cell (only covers the event window)
    rain_by_day = {}
    for p, v in zip(time_coord.points, values):
        d = time_coord.units.num2date(p)
        k = (d.year, d.month, d.day)
        rain_by_day[k] = rain_by_day.get(k, 0) + (0 if np.isnan(v) else v)
    rain = [rain_by_day.get((c.year, c.month, c.day), 0) for c in sm_time_arr[lo:hi]]

    fig, ax = plt.subplots(figsize=(8, 3.5))
    ax.plot(offsets, sm_series, 'o-', label='SM at cell')
    ax.axvline(0,  color='k', ls='--', label='first wet day')
    ax.axvline(-1, color='r', ls='--', label='day used')
    ax.set_xlabel('days from first wet day'); ax.set_ylabel('SM')
    ax2 = ax.twinx(); ax2.bar(offsets, rain, alpha=0.3, color='tab:blue')
    ax2.set_ylabel('rain at cell (event window)')
    ax.legend(); plt.show()
```

```python
for catchment_num in ['23']:
    
    print(f"Running for {catchment_num}")
    if not os.path.exists(os.path.join(OUT_DIR, f"Catchment_{catchment_num}")):
        os.mkdir(os.path.join(OUT_DIR, f"Catchment_{catchment_num}"))
    start_catchment_time = time.time()

    catchment_name  = CATCHMENT_LOOKUP_DICT[str(catchment_num)]
    # print(f"Starting: {catchment_name}")
    combined_fp = os.path.join(OUT_DIR, f"Catchment_{catchment_num}", f"{catchment_name}_new.pkl")
    
    if not os.path.exists(combined_fp):
        print(f"Already exists for {catchment_name}")
    else:
        boundary_gdf    = CATCHMENTS[CATCHMENTS['HA_NUM'] == str(catchment_num)]
        CATCHMENT_POLY = boundary_gdf.geometry.iloc[0]

        results = []
        
        # --------------------------
        #  For creating mask 
        # --------------------------
        begin_time = time.time()
        full_rain_cube = get_rainfall_cube_subsection(2015, '01', f"/scratch/hydro5/users/ld14116/SDM_bias_correction/Hourly/01/", 1, 2)
        FULL_MASK_2D   = mask_cube_with_catchment_full_grid(full_rain_cube[0], CATCHMENT_POLY, method='center_point')
        year_cube, mask_x_offset, mask_y_offset = subset_cube_to_bbox(full_rain_cube, CATCHMENT_POLY, buffer=0)
        ny_sub = year_cube.shape[1]
        nx_sub = year_cube.shape[2]
        mask_2d_sub = FULL_MASK_2D[mask_y_offset:mask_y_offset + ny_sub, 
                                    mask_x_offset:mask_x_offset + nx_sub]
        # print(f"Mask created in {time.time()-begin_time:.1f}s")
        del full_rain_cube, year_cube

        flood_dir    = f"/scratch/hydro4/users/kv25483/FutureFlood/Data/PluvialResults/5km_total/Catchment_{catchment_num}/"
        flood_numpy = {}
        flood_cubes = {}
        current_ens = None
        DIAGNOSE_EVENT = [1,2,3,4,5]
        n_days_ls      = [1]
        rain_n_days_ls = [1] 
        all_results    = []

        for ens_num in ENSEMBLE_MEMBERS:

            fp = os.path.join(OUT_DIR,f"Catchment_{catchment_num}",f"{catchment_name}_EM{ens_num}_new.pkl")

            if os.path.exists(fp):
                print(f"Loading {catchment_name}, EM{ens_num}")
                df = pd.read_pickle(fp)
                results.append(df)                

            else:
                start_time_ens = time.time()
                results_this_ens = []
                
                # --------------------------
                # Get rainfall event data
                # --------------------------
                rainfall_events  = pd.read_csv(RAINFALL_CSV_DIR + f"{catchment_name}_{ens_num}_full_events_with_event_nums.csv")
                rainfall_events['event_num']  = range(1, len(rainfall_events) + 1)
                rainfall_events['hydro_year'] = rainfall_events['start_year'].where(rainfall_events['start_month'] != 12,
                        rainfall_events['start_year'] + 1)

                rainfall_cube_dir = f"/scratch/hydro5/users/ld14116/SDM_bias_correction/Hourly/{ens_num}/"
                print(f"EM{ens_num}: {len(rainfall_events)} events to process")
                
                # --------------------------
                # Get dictionary containing a dictionary of results for each events
                # -------------------------
                event_details_cache = {
                    ev: get_rainfall_event_details(rainfall_events, ev)
                    for ev in rainfall_events['event_num']}
                
                # --------------------------
                # Loop through 'hydro-years'
                # -------------------------
                for year, events_in_year in rainfall_events.groupby('hydro_year'):
                                    
                    # --------------------------
                    # Get soil moisture data for this 'hydro-year' (which can be shared among events)
                    # -------------------------
                    sm_dir  = f'/scratch/hydro4/shared_data/climate_projections/UKCP18/UKCP_local/Soil_moisture/5km_regridded/Ens_{ens_num}/'
                    sm_file = glob.glob(f"{sm_dir}/r001i1p*****_{year-1}1201-{year}1130_mrso.nc")[0]
                    sm      = iris.load(sm_file)[0]

                    needs_extra_sm = ((events_in_year['start_month'] == 12) & (events_in_year['start_day'] == 1)).any()

                    if needs_extra_sm:
                        prev_sm_file = glob.glob(f"{sm_dir}/r001i1p*****_{year-2}1201-{year-1}1130_mrso.nc")
                        if prev_sm_file:
                            prev_sm = iris.load(prev_sm_file[0])[0]
                            prev_sm.attributes = {}
                            sm.attributes = {}
                            sm = iris.cube.CubeList([prev_sm, sm]).concatenate_cube()
                            print(f"  Loaded extra SM file for year {year-2}-{year-1} (early-December event)")
                        else:
                            print(f"  WARNING: needed prior SM file for {year-2}-{year-1} but not found")

                    sm.data      = np.where(FULL_MASK_2D, sm.data, np.nan)
                    sm           = filter_closer_to_catchment(sm, CATCHMENT_POLY, plot=False)
                    sm_data      = sm.data
                    sm_time_arr  = np.array([cell.point for cell in sm.coord('time').cells()])
                    # Build a date-to-index dictionary 
                    sm_day_idx = {(c.year, c.month, c.day): i for i, c in enumerate(sm_time_arr)}
                    
                    # --------------------------
                    # Loop through all events within this 'hydro-year'
                    # -------------------------
                    for row in events_in_year.itertuples():
                        
                        # -------------------------
                        # Get details for this event from the dictionary
                        # -------------------------
                        event_num     = int(row.event_num)
                        event_details = event_details_cache[event_num]
                        
                        # -------------------------
                        # Get rainfall data for the hydro-year this event happened in
                        # -------------------------
                        year_cube = get_rainfall_cube_subsection(year, ens_num, rainfall_cube_dir, 
                                                             event_details['start_idx']-1, event_details['stop_idx'])
                        year_cube.data = np.where(FULL_MASK_2D, year_cube.data, np.nan)
                        year_cube, x_offset, y_offset = subset_cube_to_bbox(year_cube, CATCHMENT_POLY, buffer=0)
                        time_coord  = year_cube.coord('time')

                        # -------------------------
                        # Extract the location of the cell where the peak occurred
                        # -------------------------        
                        this_event_results = find_max_precip_location_new(
                            year_cube, event_details['start_idx'], event_details['stop_idx'],
                            x_offset=x_offset, y_offset=y_offset, mask_2d=mask_2d_sub)                        
                        

                        # -------------------------
                        # Get rainfall data for the hydro-year this event happened in
                        # -------------------------                        
                        t0 = time_coord.units.num2date(np.floor(time_coord.points[0]))
                        t_end = time_coord.units.num2date(time_coord.points[-1])
                        event_start = cftime.Datetime360Day(event_details['start_year'], 
                                                     event_details['start_month'], 
                                                     event_details['start_day'], 
                                                     event_details['start_hour'])
                        # Check timings make sense
                        event_start_within_cube = t0 <= event_start <= t_end
                        if not event_start_within_cube:
                            this_event_results['time_mismatch'] = True
                            print("TIME MISMACH")
                        else:
                            this_event_results['time_mismatch'] = False                        

                        # -------------------------
                        # Add rainfall event metadata to this_event_results from event_details dict
                        # -------------------------    
                        this_event_results.update({"ens": ens_num,
                            **{k: event_details[k] for k in [
                                "start_idx", "stop_idx", "max_precip_from_csv",
                                "start_year", "start_month", "start_day", "start_hour",
                                "event_num", "event_durations"]}})                        
                        
                        # -------------------------
                        # Extract the rainfall data for location where the peak occurred
                        # -------------------------        
                        rainfall_at_peak  = get_data_at_peak_cell(year_cube, this_event_results, 'x_idx', 'y_idx')
                        
                        # Calculate rainfall peak date
                        t = time_coord.units.num2date(time_coord.points[this_event_results["t_local"]])
                        this_event_results["rainfall_peak_day"] = cftime.Datetime360Day(t.year, t.month, t.day)

                        this_event_results.update({"hydro_day_360": (t.month % 12) * 30 + t.day,"day_360": (t.month - 1) * 30 + t.day,
                            "t_local_rebased": this_event_results["t_global"] - this_event_results["start_idx"]})

                        # Check rainfall peak consistency
                        cube_precip = float(np.float32(this_event_results["max_precip"]))
                        csv_precip = float(np.float32(this_event_results["max_precip_from_csv"]))

                        this_event_results["mismatch"] = not math.isclose(cube_precip, csv_precip, abs_tol=0.001)
                        if this_event_results["mismatch"]:
                            print(f"MISMATCH — ens={ens_num}, year={year}, event={event_num}")
                            print(f"  cube: {cube_precip:.8f}")
                            print(f"  csv:  {csv_precip:.8f}")    
                            
                        # Get rainfall data (and trim off leading and trailing zeroes)    
                        WET_THRESH = 0.1   # define once at the top of the script

                        values  = np.ma.filled(rainfall_at_peak.data, np.nan)
                        wet_idx = np.flatnonzero(values > WET_THRESH)

                        first_wet, last_wet = int(wet_idx[0]), int(wet_idx[-1])
                        trimmed_values = values[first_wet:last_wet + 1]
                        t_onset = time_coord.units.num2date(time_coord.points[first_wet])
                        anchor_key = (t_onset.year, t_onset.month, t_onset.day)

                        this_event_results['first_wet_idx'] = first_wet
                        this_event_results['trimmed_duration_hrs'] = last_wet - first_wet + 1                        
                        
                        # -------------------------
                        # Extract the flood data for location where the peak occurred
                        # -------------------------        
                        # The flooding results contain all events for each catchment / em in one .nc file
                        # So only need to load once per ensemble member
                        if ens_num != current_ens:
                            if flood_cubes:
                                del flood_cubes, flood_numpy
                                gc.collect()
                            ens_flood_dir = f"{flood_dir}/Ens{ens_num}_{catchment_num}/"
                            flood_cubes = {}
                            flood_numpy = {}
                            for depth in [10, 30]:
                                area_cube = prepare_flood_cube(load_3d_cube(
                                    f"{ens_flood_dir}/{depth}cm/flooded_area_5km_total_Ens{ens_num}_{catchment_num}_{depth}cm.nc"))
                                area_cube.data = np.where(FULL_MASK_2D, area_cube.data, np.nan)
                                area_cube = filter_closer_to_catchment(area_cube, CATCHMENT_POLY, plot=False)
                                #print(f"Area cube shape: {area_cube.shape}")
                                vol_cube  = prepare_flood_cube(load_3d_cube(
                                    f"{ens_flood_dir}/{depth}cm/flooded_volume_5km_total_Ens{ens_num}_{catchment_num}_{depth}cm.nc"))
                                vol_cube.data = np.where(FULL_MASK_2D, vol_cube.data, np.nan)
                                vol_cube = filter_closer_to_catchment(vol_cube, CATCHMENT_POLY, plot=False)
                                flood_cubes[depth] = {'area': area_cube, 'vol': vol_cube}
                                flood_numpy[depth] = {'area': area_cube.data, 'vol': vol_cube.data}
                            current_ens = ens_num     
                        
                        # Get the location of the peak
                        xi = int(this_event_results['x_idx'])
                        yi = int(this_event_results['y_idx'])
                        
                        # Search for flood values at this location
                        flood_stats = {}
                        for depth in [10, 30]:
                            flood_stats[f"{depth}cm_area"]   = flood_numpy[depth]['area'][this_event_results['event_num'] - 1, yi, xi]
                            flood_stats[f"{depth}cm_volume"] = flood_numpy[depth]['vol'][this_event_results['event_num'] - 1, yi, xi]
                        
                        # Add to the dictionary of results for this event
                        this_event_results.update(flood_stats)                            
                            
                        # -------------------------
                        # PLotting function to sense check results
                        # -------------------------
                        maybe_diagnose(
                            this_event_results,
                            event_num=this_event_results['event_num'],
                            condition=(this_event_results['event_num'] in DIAGNOSE_EVENT),
                            sm=sm,
                            flood_cubes=flood_cubes,
                            catchment_poly=CATCHMENT_POLY,
                            ens_num=ens_num,
                            rainfall_cube_dir=rainfall_cube_dir, 
                            FULL_MASK_2D= FULL_MASK_2D,
                         year_cube = year_cube)
                        
                        # -------------------------
                        # Get soil moisture data for the day before the rainfall started
                        # -------------------------    
                        start_key = (int(event_details['start_year']),int(event_details['start_month']),
                                     int(event_details['start_day']))
                        sm_idx = sm_day_idx.get(start_key) -1 # to ensure fully the day before the event starts

                        if sm_idx is None:
                            print(f"  Warning: no SM value for event={event_num}, start={start_key}")
                            sm_at_start_field = np.full(sm_data.shape[1:], np.nan)
                        else:
                            sm_at_start_field = sm_data[sm_idx]          # (y, x) field for the start day

                        mean_sm_stats = {'sm_event_start': float(sm_at_start_field[yi, xi])}
                        
                        print("hydro year:", year, "| anchor_key:", anchor_key,
                              "| CSV start:", (event_details['start_year'], event_details['start_month'], event_details['start_day']))
                        print("SM axis:", sm_time_arr[0], "->", sm_time_arr[-1], "| n =", len(sm_time_arr))
                        print("time_mismatch:", this_event_results['time_mismatch'])                        
                        
                        assert sm_data.shape[1:] == year_cube.shape[1:], \
                            f"SM grid {sm_data.shape[1:]} != rainfall grid {year_cube.shape[1:]}"

                        sm_idx = sm_day_idx.get(anchor_key)
                        if sm_idx is None or sm_idx - 1 < 0:
                            print(f"  Warning: no SM for day before {anchor_key}, event={event_num}")
                            sm_before_field = np.full(sm_data.shape[1:], np.nan)
                        else:
                            sm_before_field = sm_data[sm_idx - 1]       # day before first wet day

                        this_event_results['sm_day_before_onset'] = float(sm_before_field[yi, xi])
                        this_event_results['sm_anchor_date'] = anchor_key
                        
                        if this_event_results['event_num'] in DIAGNOSE_EVENT and sm_idx is not None:
                                plot_sm_check(sm_data, sm_time_arr, sm_day_idx, anchor_key,
                                  yi, xi, values, time_coord, n=10)
                        
                        # -------------------------
                        # Study the conditions around the peak in space for both rainfall and flooding
                        # -------------------------    
                        if this_event_results['event_num'] in DIAGNOSE_EVENT:
                            plot=True
                        else:
                            plot = False
                            
                        peak_results = analyse_peak_event(
                            this_event_results, year_cube, flood_numpy, flood_cubes,
                            CATCHMENT_POLY, boundary_gdf, neighbourhood_size=1,
                            threshold_levels=[0.5, 0.6, 0.8],
                            plot=plot)
                        this_event_results.update(peak_results)    
                        
                        # -------------------------
                        # Add the results for this ensemble member to the results for the whole ensemble member
                        # -------------------------   
                        results_this_ens.append(this_event_results)

                    del year_cube

                pd.DataFrame(results_this_ens).to_pickle(
                    os.path.join(OUT_DIR, f"Catchment_{catchment_num}", f"{catchment_name}_EM{ens_num}_new.pkl"))
                #print(f"EM{ens_num}: saved in {round((time.time() - start_time_ens)/60,2)} minutes")
                # Add to dataframe of all results
                results.append(pd.DataFrame(results_this_ens))

        # Convert to dataframe
        results_df = pd.concat(results, ignore_index=True) 

        results_df.to_pickle(combined_fp)
        #print(f"Combined file saved: {combined_fp}")
        for ens_num in ENSEMBLE_MEMBERS:
            fp = os.path.join(OUT_DIR, f"Catchment_{catchment_num}", f"{catchment_name}_EM{ens_num}_new.pkl")
            if os.path.exists(fp):
                os.remove(fp)
        #print("Done — individual EM files cleaned up")

        print(f"Catchment {catchment_num} ran in {round((time.time() - start_catchment_time)/60,2)} minutes")
```

```python
def run_spatial_diagnostics(row, event_num, sm, flood_cubes, catchment_poly,
                             ens_num, rainfall_cube_dir, FULL_MASK_2D, year_cube, depth=10):
    
    """
    Plots a 2x2 spatial check for a single event.
    Rainfall cube is loaded here on demand and discarded after plotting.

    Parameters
    ----------
    row               : namedtuple   — single event row from itertuples()
    event_num         : int          — event number (passed explicitly)
    sm                : iris.Cube    — soil moisture cube for this (ens, year)
    HC_CUBE           : iris.Cube    — hydraulic conductivity (static)
    flood_cubes       : dict         — pre-loaded flood cubes, keyed [ens_num][depth]
    catchment_poly    : shapely geom — catchment boundary polygon
    ens_num           : str          — ensemble member ID
    rainfall_cube_dir : str          — directory for hourly rainfall cubes
    depth             : int          — flood depth threshold to plot (10 or 30)
    """
    # Load and slice rainfall cube only when needed
    rainfall_slice   = year_cube[row['t_local'], :, :]
    # rainfall_slice.data = np.where(FULL_MASK_2D, rainfall_slice.data, np.nan)
    
    sm_slice    = sm[row['hydro_day_360'], :, :]
    flood_slice = flood_cubes[depth]['area']
    flood_slice = flood_slice[event_num-1, :,:]
    
    fig, axs = plt.subplots(ncols=2, nrows=2, figsize=(10, 8),
                            subplot_kw={'projection': ccrs.OSGB()})
    axs = axs.flatten()

    plot_peak_check(axs[0], sm_slice,    catchment_poly, row, title='Soil moisture')
    plot_peak_check(axs[1], rainfall_slice, catchment_poly, row, title='Precipitation')
    # plot_peak_check(axs[2], HC_CUBE,     catchment_poly, row, title='Hydraulic conductivity')
    plot_peak_check(axs[2], flood_slice, catchment_poly, row, title=f'Flooded area ({depth}cm)')

    fig.suptitle(
        f"Spatial diagnostics — ens={ens_num}, year={row['start_year']}, event={event_num}\n"
        f"Peak day: {row['rainfall_peak_day']}",
        fontsize=12)
    plt.tight_layout()
    plt.show()

    # Discard — no reason to keep it in memory after plotting
    # del rainfall_yr_cube
    gc.collect()   

def maybe_diagnose(row, event_num, condition=False, **kwargs):
    """
    Calls run_spatial_diagnostics only if condition is True.
    event_num is passed explicitly rather than read from row.
    """
    if condition:
        run_spatial_diagnostics(row, event_num=event_num, **kwargs)
```
