```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from pyproj import Transformer
```

### Get data across all ensemble members

```python
ENS_LIST = ['01', '04', '05', '06', '07', '08', '09', '10', '11', '12', '13', '15']
all_ens = list()
for ens in ENS_LIST:
    this_en = pd.read_csv(f"../../Data/Events/results_ens_{ens}.csv")
    all_ens.append(this_en)
results_all = pd.concat(all_ens)
# results_all['30cm_fld_vol_under_threshold']
```

```python
# Define transformer (WGS84 → BNG)
transformer = Transformer.from_crs("EPSG:4326", "EPSG:27700", always_xy=True)

# Newcastle coordinates (lon, lat!)
lon, lat = -1.601382, 54.970304

newcastle_x, newcastle_y = transformer.transform(lon, lat)

# Euclidean distance in metres (BNG is in metres)
dist = np.sqrt((results_all['x_coord'] - newcastle_x)**2 +(results_all['y_coord'] - newcastle_y)**2)

# Create boolean column (within 10 km = 10,000 m)
results_all['within_10km_city_centre'] = dist <= 10000

results_all['distance_to_city_centre'] = dist
```

```python
near_newcastle = results_all[results_all["within_10km_city_centre"]==True]
```

```python
plt.scatter(near_newcastle['max_precip'],near_newcastle['10cm_fld_area_at_peak'])
plt.xlabel('Maximum precipitation intensity')
plt.ylabel('Flood area at grid cell with maximum')
```

```python
plt.scatter(results_all['total_acc'],results_all['30cm_fld_area_at_peak'])
plt.xlabel('Maximum precipitation intensity')
plt.ylabel('Flood area at grid cell with maximum')
```

```python
plt.scatter(results_all['10cm_neighbourhood_sum'],results_all['10cm_fld_area_under_cluster'])
plt.xlabel('Maximum precipitation intensity')
plt.ylabel('Flood area at grid cell with maximum')
```

```python
def plot_with_best_fit_line(ax, col1, col2, label1, label2, title):
    x = results_all[col1]
    y = results_all[col2]

    ax.scatter(x, y)

    # 🔹 Fit line
    m, b = np.polyfit(x, y, 1)

    # Predicted values
    y_pred = m * x + b

    # R²
    ss_res = np.sum((y - y_pred)**2)
    ss_tot = np.sum((y - np.mean(y))**2)
    r2 = 1 - (ss_res / ss_tot)

    # 🔹 Create line values
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = m * x_line + b

    # 🔹 Plot line
    ax.plot(x_line, y_line, color='black')

    ax.text( 0.05, 0.95,  f'$R^2 = {r2:.2f}$', transform=ax.transAxes, verticalalignment='top')

    ax.set_xlabel(label1)
    ax.set_ylabel(label2)
    
    ax.set_title(title)
```

```python
fig, axs = plt.subplots(ncols=2, figsize=(10,3))
plot_with_best_fit_line(axs[0], 'total_acc', '10cm_fld_area_under_cluster',
                       'Accumulated rainfall at grid \n cell with maximum', 'Flood area at grid \n cell with maximum',
                       'Flooding over 10cm')
plot_with_best_fit_line(axs[1], 'total_acc', '30cm_fld_area_under_cluster',
                       'Accumulated rainfall at grid \n cell with maximum', 'Flood area at grid \n cell with maximum',
                       'Flooding over 30cm')
# fig.tight_layout()
```

```python
fig, axs = plt.subplots(ncols=2, figsize=(10,3))
plot_with_best_fit_line(axs[0], 'max_precip', '10cm_fld_area_under_cluster',
                       'Maximum precipitation intensity', 'Flood area at grid \n cell with maximum',
                       'Flooding over 10cm')
plot_with_best_fit_line(axs[1], 'max_precip', '30cm_fld_area_under_cluster',
                       'Maximum precipitation intensity', 'Flood area at grid \n cell with maximum',
                       'Flooding over 30cm')
# fig.tight_layout()
```
