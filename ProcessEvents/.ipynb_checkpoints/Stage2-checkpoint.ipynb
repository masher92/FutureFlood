```python
def analyse_peak_event(row, neighbourhood_size, threshold_level, plot = False):
    # ── Parameters ────────────────────────────────────────────────────────────────
    # threshold_level: fraction of the peak value used to define the "significant"
    #   rainfall region. E.g. 0.8 means only cells with >= 80% of the peak rainfall
    #   are considered part of the same storm cluster.
    threshold_level    = threshold_level
    peak_value         = row.max_precip
    threshold_value    = peak_value * threshold_level  # absolute mm threshold for this event

    # neighbourhood_size: radius (in grid cells) of the box around the peak cell
    #   used to compute local neighbourhood rainfall. 1 = a 3x3 box, 2 = a 5x5 box.
    neighbourhood_size = neighbourhood_size

    # ── Load and slice rainfall cube ──────────────────────────────────────────────
    # Load the full hourly rainfall cube for this year/ensemble, then slice to:
    #   1. The event window: [start_idx:stop_idx] selects the hours covering this event
    #   2. The peak timestep: [t_local] selects the single hour of maximum rainfall
    # Result is a 2D spatial slice (y, x) at the moment of peak rainfall.
    start_time=time.time()
    # Only the timesteps start_idx:stop_idx are ever read from disk
    event_cube = get_rainfall_cube_subsection(row.year, row.ens, rainfall_cube_dir,
                                   start_idx=row.start_idx, stop_idx=row.stop_idx)
    event_cube.data = np.where(FULL_MASK_2D, event_cube.data, np.nan)
    event_cube, x_offset, y_offset = subset_cube_to_bbox(event_cube, CATCHMENT_POLY, buffer=0)

    # Then index t_local within that already-sliced cube
    peak_slice = event_cube[row.t_local, :, :]
    print(time.time()-start_time)

    # Extract as a float numpy array so we can safely manipulate fill values and NaNs.
    # .copy() prevents any modifications from propagating back to the iris cube.
    rain = peak_slice.data.copy().astype(float)

    # ── Clean invalid values ──────────────────────────────────────────────────────
    # Replace sentinel fill values (-99999) and implausibly large values (> 1e19,
    # which are likely unmasked fill values from the netCDF) with NaN so they are
    # excluded from all subsequent calculations.
    rain = np.where((rain == -99999) | (rain > 1e19), np.nan, rain)

    # ── Build threshold mask and label connected clusters ─────────────────────────
    # threshold_mask: True for every cell that is both:
    #   (a) >= threshold_value (i.e. >= 80% of the peak cell's rainfall), AND
    #   (b) not NaN (excludes fill/missing cells from being labelled as clusters)
    # This identifies all cells that are "meaningfully rainy" relative to the peak.
    threshold_mask = (rain >= threshold_value) & (~np.isnan(rain))

    # Use scipy.ndimage.label to assign a unique integer ID to each spatially
    # connected group of True cells in the mask.
    # diagonal=True uses an 8-connectivity structure (cells connected via corners
    #   count as neighbours), which is more physically realistic for storm clusters.
    # diagonal=False uses 4-connectivity (only N/S/E/W neighbours), which is stricter.
    diagonal = False
    if not diagonal:
        # 4-connectivity: only directly adjacent cells (cross-shaped kernel)
        labeled, num_clusters = label(threshold_mask)
    else:
        # 8-connectivity: adjacent + diagonal cells (full 3x3 kernel)
        structure = generate_binary_structure(2, 2)
        labeled, num_clusters = label(threshold_mask, structure=structure)

    # ── Extract the cluster containing the peak cell ──────────────────────────────
    # labeled[y, x] gives the cluster ID at the peak cell's grid location.
    # cluster_mask is then True only for cells belonging to that same cluster,
    # i.e. the connected storm region that contains the rainfall peak.
    # If peak_label == 0 the peak cell itself was below the threshold — worth
    # checking if this happens unexpectedly (peak_value may be very low).
    peak_label   = labeled[row.y_idx, row.x_idx]
    cluster_mask = (labeled == peak_label)

    # ── Compute neighbourhood rainfall sum around the peak cell ───────────────────
    # Define a bounding box of radius n around the peak cell, clamped to the array
    # edges so we don't go out of bounds near the domain boundary.
    # This is intentionally independent of the cluster mask above: it captures
    # the raw local rainfall context (including cells below threshold) in a fixed
    # window, rather than the shape of the storm cluster.
    n           = neighbourhood_size
    y0, y1      = max(0, row.y_idx - n), min(rain.shape[0], row.y_idx + n + 1)
    x0, x1      = max(0, row.x_idx - n), min(rain.shape[1], row.x_idx + n + 1)
    neighbourhood = rain[y0:y1, x0:x1]  # shape: up to (2n+1, 2n+1), smaller near edges

    # Sum all valid (non-NaN) cells in the neighbourhood window, then subtract the
    # peak cell's own value so the result reflects only the surrounding cells.
    # This gives a measure of how much "supporting" rainfall surrounds the peak.
    neighbourhood_sum = float(np.nansum(neighbourhood) - peak_value)

    # ── Rainfall totals ───────────────────────────────────────────────────────────
    # nansum ignores NaN cells (fill values) in each region.
    # neighbourhood uses the raw bounding box slice; cluster and threshold use
    # their boolean masks applied to the full 2D rain array.

    neighbourhood_rain_total = float(np.nansum(rain[y0:y1, x0:x1]))
    cluster_rain_total       = float(np.nansum(rain[cluster_mask]))
    threshold_rain_total     = float(np.nansum(rain[threshold_mask]))

    # ── Flood totals (if you have a 2D flood numpy array) ────────────────────────
    # Use the same masks — they are all in the same grid space so can be applied
    # directly to the flood array. Make sure to use the correct event slice first.
    flood_2d = flood_numpy[10]['area'][row.event_num_this_ens - 1, :, :]  # shape (y, x)

    neighbourhood_flood_total = float(np.nansum(flood_2d[y0:y1, x0:x1]))
    cluster_flood_total       = float(np.nansum(flood_2d[cluster_mask]))
    threshold_flood_total     = float(np.nansum(flood_2d[threshold_mask]))

    # ── Cell counts (useful for normalising or sanity checking) ───────────────────
    # Tells you how many valid (non-NaN) cells are in each region
    neighbourhood_n_cells = int(np.sum(~np.isnan(rain[y0:y1, x0:x1])))
    cluster_n_cells       = int(np.sum(cluster_mask))
    threshold_n_cells     = int(np.sum(threshold_mask))

    print(f"Neighbourhood — rain: {neighbourhood_rain_total:.2f} mm, "
          f"flood: {neighbourhood_flood_total:.2f}, cells: {neighbourhood_n_cells}")
    print(f"Cluster       — rain: {cluster_rain_total:.2f} mm, "
          f"flood: {cluster_flood_total:.2f}, cells: {cluster_n_cells}")
    print(f"Threshold     — rain: {threshold_rain_total:.2f} mm, "
          f"flood: {threshold_flood_total:.2f}, cells: {threshold_n_cells}")

    test = {'neighbourhood_rain_total': neighbourhood_rain_total,
    'cluster_rain_total':       cluster_rain_total,
    'threshold_rain_total':     threshold_rain_total,
    'neighbourhood_flood_total': neighbourhood_flood_total,
    'cluster_flood_total':       cluster_flood_total,
    'threshold_flood_total':     threshold_flood_total,
    'neighbourhood_n_cells':    neighbourhood_n_cells,
    'cluster_n_cells':          cluster_n_cells,
    'threshold_n_cells':        threshold_n_cells}
    
    if plot ==True:
        plot_cluster_check(
        rain=rain,
        peak_slice=peak_slice,
        threshold_mask=threshold_mask,
        cluster_mask=cluster_mask,
        row=row,
        y0=y0, y1=y1, x0=x0, x1=x1,
        threshold_level=threshold_level,
        neighbourhood_sum=neighbourhood_sum,
        neighbourhood_size = neighbourhood_size,
        flood=flood_cubes[10]['area'][row.event_num_this_ens - 1],  # optional
        catchment_poly=CATCHMENT_POLY,
        boundary_gdf=boundary_gdf,
    )
    

    return test
```

```python
import pandas as pd
import numpy as np
# import geopandas as gpd
import os
import re
import iris
import cftime 
import gc
import glob
# import cartopy.crs as ccrs
# import matplotlib.pyplot as plt
import datetime
import time 
from scipy.ndimage import label, generate_binary_structure

from functions import (get_rainfall_cube, load_3d_cube, prepare_flood_cube, filter_closer_to_catchment, subset_cube_to_bbox)
from functions_stage2 import (maybe_diagnose, get_data_at_peak_cell, plot_peak_check,get_rainfall_cube_subsection, 
                    plot_cluster_check)
from config import CATCHMENT_LOOKUP_DICT, OUT_DIR, CATCHMENTS # MOLLY_DIR_FF, RAINFALL_CSV_DIR, , ENSEMBLE_MEMBERS,  

# Get list of catchments to run
all_catchments = set(CATCHMENT_LOOKUP_DICT.keys())
files = os.listdir(OUT_DIR)
completed_catchments = {re.search(r'Catchment_(.+)', f).group(1) for f in files if re.search(r'Catchment_(.+)', f)}
catchments_to_run = (all_catchments - completed_catchments) - {105} - {14}
```

```python
def apply_mask_to_cube(cube, mask_2d):
    """Apply a prebuilt 2D boolean mask to a cube, returning a masked copy."""
    cube_masked = cube.copy()
    cube_masked.data = np.where(mask_2d[np.newaxis, :, :], cube.data, np.nan)
    return cube_masked
```

```python
catchment_num = "23"

catchment_name = CATCHMENT_LOOKUP_DICT[str(catchment_num)]
boundary_gdf   = CATCHMENTS[CATCHMENTS['HA_NUM'] == str(catchment_num)]
CATCHMENT_POLY = boundary_gdf.geometry.iloc[0]

# Get rainfall event data and apply pre-processing
rainfall_events = pd.read_pickle(f"../../Data/EventDetails/Catchment_{catchment_num}/{catchment_name}.pkl")

#      ## Do some value manipulation
#     rainfall_events['values'] = rainfall_events['values'].apply(
#         lambda x: np.fromstring(x.strip('[]'), sep=' '))
if catchment_num == "23":
    rainfall_events['rainfall_peak_day'] = pd.to_datetime(
        rainfall_events['rainfall_peak_day'], format="%d/%m/%Y %H:%M")
else:
    rainfall_events['rainfall_peak_day'] = pd.to_datetime(
        rainfall_events['rainfall_peak_day'], format="%Y-%m-%d %H:%M:%S")

rainfall_events['rainfall_peak_360'] = rainfall_events['rainfall_peak_day'].apply(
    lambda d: cftime.Datetime360Day(d.year, d.month, d.day, d.hour, d.minute, d.second))
rainfall_events['hydro_day_360'] = (
    (rainfall_events['rainfall_peak_day'].dt.month % 12) * 30
    + rainfall_events['rainfall_peak_day'].dt.day)
rainfall_events['day_360'] = (
    (rainfall_events['rainfall_peak_day'].dt.month - 1) * 30
    + rainfall_events['rainfall_peak_day'].dt.day)
rainfall_events['t_local']    = rainfall_events['t_global'] - rainfall_events['start_idx']
rainfall_events['event_num']  = range(1, len(rainfall_events) + 1)
rainfall_events['event_num_this_ens'] = rainfall_events.groupby('ens').cumcount() + 1

# Get hydraulic conductivity data
HC_CUBE      = iris.load(f"/scratch/hydro4/users/kv25483/FutureFlood/Data/HydraulicConductivity/5km_{catchment_num}.nc")[0]
HC_CUBE.data = np.where(FULL_MASK_2D, HC_CUBE.data, np.nan)
HC_CUBE = filter_closer_to_catchment(HC_CUBE, CATCHMENT_POLY, plot=False)
HC_DATA      = HC_CUBE.data  # realise once — shape (y, x)
flood_dir    = f"/scratch/hydro4/users/kv25483/FutureFlood/Data/PluvialResults/5km_total/Catchment_{catchment_num}/"

n_days_ls      = [2, 3, 4, 5]
all_results    = []
DIAGNOSE_EVENT = []

flood_numpy = {}  # will hold realised numpy arrays, keyed [depth]['area'/'vol']
flood_cubes = {}  # still needed for maybe_diagnose (expects iris cubes)
current_ens = None

    
start_time1 = time.time()
full_rain_cube = get_rainfall_cube(2015, '01', f"/scratch/hydro5/users/ld14116/SDM_bias_correction/Hourly/01/")   # any year, grid is same
FULL_MASK_2D = mask_cube_with_catchment_full_grid(full_rain_cube[0], CATCHMENT_POLY, method='full_cell')
print(f"Creating a mask for EM{ens_num} in {round(time.time() - start_time1, 2)}s")   

for (ens_num, year), group in rainfall_events[1:5].groupby(['ens', 'year']):

    rainfall_cube_dir = f"/scratch/hydro5/users/ld14116/SDM_bias_correction/Hourly/{ens_num}/"
    print(f"Processing ens={ens_num}, year={year} ({len(group)} events)")

    # ── Reload flood cubes only when ensemble member changes ──────────────────
    if ens_num != current_ens:
        if flood_cubes:
            del flood_cubes, flood_numpy
            gc.collect()
        ens_flood_dir = f"{flood_dir}/Ens{ens_num}_{catchment_num}/"
        flood_cubes = {}
        flood_numpy = {}
        for depth in [10, 30]:
            area_cube = prepare_flood_cube(load_3d_cube(
                f"{ens_flood_dir}/{depth}cm/flooded_area_5km_total_Ens{ens_num}_{catchment_num}_{depth}cm.nc"))
            area_cube.data = np.where(FULL_MASK_2D, area_cube.data, np.nan)
            area_cube = filter_closer_to_catchment(area_cube, CATCHMENT_POLY, plot=False)
            vol_cube  = prepare_flood_cube(load_3d_cube(
                f"{ens_flood_dir}/{depth}cm/flooded_volume_5km_total_Ens{ens_num}_{catchment_num}_{depth}cm.nc"))
            vol_cube.data = np.where(FULL_MASK_2D, vol_cube.data, np.nan)
            vol_cube = filter_closer_to_catchment(vol_cube, CATCHMENT_POLY, plot=False)
            flood_cubes[depth] = {'area': area_cube, 'vol': vol_cube}  # iris, for diagnostics
            flood_numpy[depth] = {'area': area_cube.data, 'vol': vol_cube.data}  # numpy, for stats
        current_ens = ens_num

    # ── Load SM cube and realise to numpy once per (ens, year) ───────────────
    sm_dir  = f'/scratch/hydro4/shared_data/climate_projections/UKCP18/UKCP_local/Soil_moisture/5km_regridded/Ens_{ens_num}/'
    sm_file = glob.glob(f"{sm_dir}/r001i1p*****_{year-1}1201-{year}1130_mrso.nc")[0]
    sm      = iris.load(sm_file)[0]
    sm.data = np.where(FULL_MASK_2D, sm.data, np.nan)
    sm = filter_closer_to_catchment(sm, CATCHMENT_POLY, plot=False)
    sm_data = sm.data  # realise once — shape (time, y, x)

    # Build time array once per group (only needs iris, done outside row loop)
    sm_example  = get_data_at_peak_cell(sm, group.iloc[0], 'x_idx', 'y_idx')
    sm_time_arr = np.array([cell.point for cell in sm_example.coord('time').cells()])

    start_time = time.time()

    for row in group.itertuples():

        # ── Diagnostics (still uses iris cubes for plotting) ──────────────────
        maybe_diagnose(
            row,
            event_num=row.event_num,
            condition=(row.event_num in DIAGNOSE_EVENT),
            sm=sm,
            HC_CUBE=HC_CUBE,
            flood_cubes=flood_cubes,
            catchment_poly=CATCHMENT_POLY,
            ens_num=ens_num,
            rainfall_cube_dir=rainfall_cube_dir)

        xi = int(row.x_idx)
        yi = int(row.y_idx)

        # ── SM stats — direct numpy indexing, no iris overhead ────────────────
        sm_at_peak_data   = sm_data[:, yi, xi]
        rainfall_peak_360 = row.rainfall_peak_360

        mean_sm_stats = {}
        for n_days in n_days_ls:
            delta       = datetime.timedelta(days=n_days)
            mask        = (sm_time_arr >= (rainfall_peak_360 - delta)) & (sm_time_arr < rainfall_peak_360)
            subset_data = sm_at_peak_data[mask]

            if subset_data.size == 0:
                print(f"  Warning: empty SM window for event={row.event_num}, "
                      f"n_days={n_days}, peak={rainfall_peak_360}")
                mean_sm_stats[f'mean_sm_{n_days}_before_event'] = float('nan')
                continue

            mean_sm_stats[f'mean_sm_{n_days}_before_event'] = float(subset_data.mean())

        # ── Flood stats — direct numpy indexing ───────────────────────────────
        flood_stats = {}
        for depth in [10, 30]:
            flood_stats[f"{depth}cm_area"]   = flood_numpy[depth]['area'][row.event_num_this_ens - 1, yi, xi]
            flood_stats[f"{depth}cm_volume"] = flood_numpy[depth]['vol'][row.event_num_this_ens - 1, yi, xi]

        # ── HC stats — direct numpy indexing ─────────────────────────────────
        hc_stats = {'hc_at_peak': HC_DATA[yi, xi]}
        
        # ------------
        test = analyse_peak_event(row, neighbourhood_size = 1, threshold_level = 0.3, plot=True)

        all_results.append({**flood_stats, **mean_sm_stats, **hc_stats, **test})

    print(f"  Took: {round(time.time() - start_time, 2)}s")

    del sm, sm_data
    gc.collect()
    
    
results_df = pd.DataFrame(all_results)
```
